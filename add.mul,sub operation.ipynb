```python
%pip install tensorflow
import tensorflow as tf

# Define some constant tensors
a = tf.constant(5.0)
b = tf.constant(3.0)

# Basic arithmetic operations
add = tf.add(a, b)
sub = tf.subtract(a, b)
mul = tf.multiply(a, b)

print(f"Tensor a: {add}")
print(f"Tensor b: {b.numpy()}")
print(f"Addition (a + b): {add.numpy()}")
print(f"Subtraction (a - b): {sub.numpy()}")
print(f"Element-wise Multiplication (a * b): {mul.numpy()}")
```
